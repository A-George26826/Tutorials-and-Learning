```python
import torch
import torch.nn as nn
import torch.optim as optim
import numpy
import pandas
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler
```

```python
X, y = load_breast_cancer(return_X_y=True,as_frame=True)
X = X.astype("float32")
```

```python
features = X.columns.to_list()
features
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=.8, random_state=42)
X_train
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
y_train.shape
```

```python
X_train = torch.FloatTensor(X_train)
y_train = torch.FloatTensor(y_train.values).unsqueeze(1)
X_test = torch.FloatTensor(X_test)
y_test = torch.FloatTensor(y_test.values).unsqueeze(1)
```

```python
X_test
```

```python
model = nn.Sequential(
    nn.Linear(30, 8),
    nn.ReLU(),
    nn.Linear(8, 4),
    nn.ReLU(),
    nn.Linear(4, 1)
)

criterion = nn.BCEWithLogitsLoss()
optimizer = optim.Adam(model.parameters(), .01)
epochs = 100
```

```python
for epoch in range(epochs):

    optimizer.zero_grad()

    preds = model(X_train)
    loss = criterion(preds, y_train)

    loss.backward()
    optimizer.step()
    print(f"Epoch: {epoch+1} ----- Loss: {loss.item()}")
```

```python
with torch.no_grad():
    test_preds = model(X_test)
    print(criterion(test_preds, y_test))
        
    predictions = (test_preds >= 0).float()

    correct = (predictions == y_test).float().sum()

    accuracy = correct / y_test.shape[0]

    accuracy_pct = accuracy.item() * 100
    print(accuracy_pct)
    
```

```python
model2 = nn.Sequential(
    nn.Linear(30, 8),
    nn.ReLU(),
    nn.Linear(8, 4),
    nn.ReLU(),
    nn.Linear(4, 1),
)

criterion2 = nn.BCEWithLogitsLoss()
optimizer2 = optim.RMSprop(model2.parameters(), .01)
```

```python

for epoch in range(epochs):

    optimizer2.zero_grad(set_to_none=True)

    logits = model2(X_train)
    loss = criterion2(logits, y_train)

    loss.backward()
    optimizer2.step()

    print(f'Epoch: {epoch+1} ---- Loss: {loss.item()}')
```

```python
with torch.no_grad():
    test_preds = model2(X_test)
    print(criterion2(test_preds, y_test))
        
    predictions = (test_preds >= 0).float()

    correct = (predictions == y_test).float().sum()

    accuracy = correct / y_test.shape[0]

    accuracy_pct = accuracy.item() * 100
    print(accuracy_pct)
```
