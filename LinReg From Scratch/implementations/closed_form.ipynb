```python
import numpy as np
```

```python
class ClosedLinReg():

    def __init__(self):
        self.coefs_ = None
        self.weights_ = None
        self.bias_ = 0.0
    
    def fit(self, X, y):
        X_full = np.c_[np.ones(X.shape[0]), X]
        self.coefs_ = np.linalg.inv(X_full.T @ X_full) @ X_full.T @ y
        self.weights_ = self.coefs_[1:]
        self.bias_ = self.coefs_[0]
    
    def predict(self, X, y):
        X_full = np.c_[np.ones(X.shape[0]), X]

        return X_full @ self.coefs_

        
```

```python
from sklearn.datasets import fetch_california_housing
from sklearn.metrics import r2_score, mean_absolute_error
```

```python
X, y = fetch_california_housing(return_X_y=True)
type(X)
```

```python
X.shape, y.shape
```

```python
y.dtype
```

```python
model = ClosedLinReg()
model.fit(X, y)
model.weights_, model.bias_
```

```python
y_preds = model.predict(X,y)
y_preds
```

```python
r2_score(y, y_preds)
```

```python
mean_absolute_error(y, y_preds)
```
